### Drop split images from OG nonsplit DF

```python
# 1. Create a set of the unique (rep, mouse_n) tuples to remove
keys_to_remove = set(zip(split_df['rep_N'], split_df['mouse_N']))
print(keys_to_remove)
# 2. Define a function to check if a row is in that set
# (Assumes nonsplit_df already has 'rep' and 'mouse_n' columns. 
#  If not, you need to extract them first using the same code as before.)
def is_bad_row(row):
    return (row['rep_N'], row['mouse_N']) in keys_to_remove

# 3. Filter nonsplit_df keeping only rows that are NOT in the set
# The '~' negates the boolean mask (Keep if NOT is_bad_row)
nonsplit_df = nonsplit_df[~nonsplit_df.apply(is_bad_row, axis=1)].copy()
```

```python
nonsplit_df
```

## Add mouse and rep number

```python
zarr_fn_pattern = "/mnt/OPERA3/Nathan/data/macrohet/mtb_tissue_localisation/data/whole_slice/rep*/mouse_*/zarr/*.zarr"
```

```python
nonsplit_zarr_fns = [fn for fn in glob.glob(zarr_fn_pattern) if ('top' not in fn) 
                                                              & ('bot' not in fn) 
                                                              & ('left' not in fn) 
                                                              & ('right' not in fn)
                                                              & ('corrupted' not in fn)
                                                              & ('jnotebook' not in fn)]
```

```python
fn = nonsplit_zarr_fns[0]
```

```python
# --- 1. Build the Lookup Dictionary ---
meta_lookup = {}

for fn in nonsplit_zarr_fns:
    # Matches: /rep2/mouse_8/ or /rep2/mouse8/
    match = re.search(r'/rep(\d+)/mouse_?(\d+)/', fn)
    if match:
        rep = int(match.group(1))
        mouse = int(match.group(2))
        basename = os.path.basename(fn)
        meta_lookup[basename] = (rep, mouse)

# --- 2. Apply to DataFrame ---
# map() looks up every zarr_address in the dictionary efficiently
mapped_values = nonsplit_df['zarr_address'].map(meta_lookup)

# Extract the values into new columns
nonsplit_df['rep_N'] = mapped_values.str[0]    # The first item in the tuple (rep)
nonsplit_df['mouse_N'] = mapped_values.str[1]  # The second item in the tuple (mouse)
nonsplit_df['split'] = None                    # Set split to None
```

### Add condition metadata

```python
metadata =pd.read_pickle('/mnt/OPERA3/Nathan/data/macrohet/mtb_tissue_localisation/results/metadata.pkl')
metadata
```

```python
# 1. Define the specific new rows you want to add
new_data = [
    # Rep 2, Mouse 11 is RIF
    {'replicate': 2, 'mouse_num': 11, 'condition': 'RIF'}
]

# 2. Add Rep 2, Mouse > 11 as PZA
# Change (12, 16) to whatever your actual max mouse number is (e.g., up to 20?)
# range(12, 16) produces: 12, 13, 14, 15
for i in range(12, 17): 
    new_data.append({'replicate': 2, 'mouse_num': i, 'condition': 'PZA'})

# 3. Create a DataFrame from the new rows
new_df = pd.DataFrame(new_data)

# 4. Append to your existing metadata
# ignore_index=True resets the index so it's continuous
metadata = pd.concat([metadata, new_df], ignore_index=True)
```

```python
metadata
```

```python
# 1. Drop 'condition' from your main df if it already exists.
#    This prevents the creation of 'condition_x' and 'condition_y'.
df = df.drop(columns=['condition'])

# 2. Perform the Left Merge
#    We assume 'metadata' is the master list, so we pull 'condition' from there.
df = df.merge(
    metadata[['replicate', 'mouse_num', 'condition']], 
    left_on=['rep_N', 'mouse_N'],        # Keys in main df
    right_on=['replicate', 'mouse_num'], # Keys in metadata
    how='left'
)

# 3. Clean up the extra keys
#    The merge added 'replicate' and 'mouse_num'; we drop them to keep df clean.
df = df.drop(columns=['replicate', 'mouse_num'])
```
