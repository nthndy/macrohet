```python
import os
import re

src_fns = glob.glob('/Volumes/OPERA3/Nathan/data/macrohet/mtb_tissue_localisation/data/whole_slice/rep*/temp_max_proj/*.zarr')
src_fns = natsorted([fn for fn in src_fns if ('top' in fn) or ('bot' in fn) or ('left' in fn) or ('right' in fn)])

# Regex to capture the mouse number and the label.
basename_pattern = re.compile(r'(\d+)_?(top|bot|left|right)\.ome\.zarr$')

# Define the common root of the destination path
DEST_ROOT = '/Volumes/OPERA3/Nathan/data/macrohet/mtb_tissue_localisation/data/whole_slice'

for src_path in src_fns:
    # 1. Extract rep_number
    path_parts = src_path.split(os.sep)
    rep_number = next((
        part.replace('rep', '') for part in path_parts if part.startswith('rep') and part[-1].isdigit()
    ), 'Unknown')

    # 2. Extract basename, mouse_number, and label
    basename = os.path.basename(src_path)
    match = basename_pattern.search(basename)

    mouse_number = 'Unknown'
    label = 'Unknown'

    if match:
        mouse_number = match.group(1)
        label = match.group(2)
        
    # Skip processing if critical information is missing
    if rep_number == 'Unknown' or mouse_number == 'Unknown' or label == 'Unknown':
        print(f"ERROR: Could not extract all required variables for {src_path}. Skipping.")
        continue

    # 3. Construct the new, standardized destination basename
    # We use .zarr as the extension instead of .ome.zarr as requested: repN_mouseN_label.zarr
    dest_basename = f'rep{rep_number}_mouse{mouse_number}_{label}.zarr'
    
    # 4. Construct the destination path
    # The destination directory structure is: DEST_ROOT/rep{rep_number}/mouse_{mouse_number}/zarr/
    dest_dir = os.path.join(
        DEST_ROOT,
        f'rep{rep_number}',
        f'mouse_{mouse_number}',
        'zarr'
    )
    
    # dest_fn is the full path to the new, renamed Zarr directory
    dest_fn = os.path.join(dest_dir, dest_basename)
    
    # 5. Create the destination directory if it doesn't exist
    os.makedirs(dest_dir, exist_ok=True)

    # 6. Copy the Zarr directory (and all subdirs) with the new name
    print(f"Copying '{src_path}' to new name '{dest_fn}'...")
    
    # # shutil.copytree is used for copying directories recursively (including contents and subdirectories)
    # if os.path.exists(dest_fn):
    #     print(f"WARNING: Destination directory already exists at '{dest_fn}'. Skipping copy.")
    # else:
    #     try:
    #         # Copy the source directory (Zarr) to the destination path with the new basename
    #         shutil.copytree(src_path, dest_fn)
    #         print("Copy and rename complete.")
    #     except Exception as e:
    #         print(f"CRITICAL ERROR during copy: {e}")

    print("---")
```
