# Interactive Single-Cell Growth Annotator
This notebook demonstrates the `GlimpseAnnotator`, a Napari-based GUI tool used to manually verify, classify, and split intracellular Mtb tracking data. 

To run this tool, you must have a single cell dataframe and corresponding raw `.mp4` video glimpses stored in `../data/glimpses/`. 
We provide a couple of single-cell example glimpses with a separate single-cell dataframe with corresponding data to work the demo. 

```python
import pandas as pd

from macrohet.single_cell_validation import GlimpseAnnotator
```

### Annotator Controls
When the Napari viewer is active, use the following keybindings to curate the data:

| Key | Action | Description |
|:---:|:---|:---|
| **N / B** | Next / Previous | Navigate through the provided ID list. |
| **G** | Mark Growth | Standard intracellular replication. |
| **U / T** | Uptake / Transfer | Extrinsic Mtb increases from the environment/neighbors. |
| **J** | Mark Junk | Flag for deletion (e.g., poor segmentation, artifacts). |
| **K** | Split Track | Divide the track at the current time-slider frame. |
| **S** | Save | Write current annotations to the curated pickle file. |

> **Note:** The `GlimpseAnnotator` automatically skips IDs if the corresponding `.mp4` file is not found in the glimpse directory.

### Load single cell dataframe

```python
# # 1. Load canonical dataset
# df = pd.read_parquet('../data/single_cell_df.parquet')

# Load example single cell dataset with corresponding glimpse mp4 videos in data folder
df = pd.read_parquet('../data/single_cell_df_validation_example.parquet')
```

### Run the trajectory annotator

```python
# This shows how the filter is applied across the entire study
# all_to_validate = df[df['r2'] < 0.7]['ID'].unique()
# print(f"Total trajectories requiring validation: {len(all_to_validate)}")

# --- 2. Minimal Working Example (MWE) ---
# The following IDs have corresponding video glimpses in the example data folder.
# These examples illustrate normal growth, slow growth, and extrinsic jumps.
example_ids = ['330.3.5.PS0000', '366.3.5.PS0000', '576.4.5.PS0000']

# --- 3. Launch Validator ---
app = GlimpseAnnotator(
    full_df=df, 
    id_list=example_ids, 
    glimpse_dir='../data/example_glimpses/', 
    save_path='../data/curated_trajectories.pkl'
)
```
