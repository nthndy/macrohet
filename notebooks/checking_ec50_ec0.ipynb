#### Functions to apply LOWESS growth model to raw data

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from sklearn.metrics import r2_score
from tqdm.auto import tqdm

# --- 1. RELAXED SMOOTHING FUNCTIONS ---

def smooth_and_fix_relaxed(area_series, window=10, spike_threshold=2.0):
    """
    Smooths data using a CENTERED rolling mean to detect spikes.
    Allows for large biological jumps (10x) but removes single-pixel noise.
    """
    # 1. Capture the original index so we can restore it later (CRITICAL FIX)
    original_index = area_series.index
    
    # 2. Reset index for calculation (so iloc[i] works 0..N)
    area_series = area_series.reset_index(drop=True)
    
    # center=True allows the mean to 'see' the jump coming
    rolling_mean = area_series.rolling(window=window, min_periods=10, center=False).mean()
    
    cleaned = area_series.copy()
    for i in range(1, len(cleaned) - 1):
        # Relaxed Threshold Check
        if cleaned.iloc[i] > spike_threshold * rolling_mean.iloc[i]:
            cleaned.iloc[i] = np.nan
        # Zero-Bounce Check
        elif (
            cleaned.iloc[i] == 0
            and cleaned.iloc[i - 1] > 0
            and cleaned.iloc[i + 1] > 0
        ):
            cleaned.iloc[i] = np.nan
            
    # 3. Interpolate to fill gaps
    result = cleaned.interpolate(limit_direction='both')
    
    # 4. Restore the original index before returning (CRITICAL FIX)
    result.index = original_index
    
    return result

def process_mtb_area_relaxed(df, window=10, spike_threshold=2.0):
    """
    Applies relaxed smoothing to the entire DataFrame with a progress bar.
    """
    df = df.copy()
    
    # Initialize tqdm for pandas
    tqdm.pandas(desc="Step 1: Smoothing Data")
    
    # Use progress_transform
    cleaned_series = df.groupby("ID")["Mtb Area (\u00b5m)"].progress_transform(
        lambda x: smooth_and_fix_relaxed(x, window, spike_threshold)
    )
    
    df["Mtb Area Processed (\u00b5m)"] = cleaned_series
    return df


# --- 2. ROBUST LOWESS FITTING ---

def fit_lowess(df, frac=0.25):
    """
    Fits Lowess model. Uses explicit index matching to prevent 
    'Must have equal len keys' errors if rows are dropped.
    """
    df = df.copy()
    # Ensure columns exist
    cols = ["Time Model (hours)", "Mtb Area Model (\u00b5m)", "r2"]
    for c in cols:
        if c not in df.columns: df[c] = np.nan

    for ID in tqdm(df["ID"].unique(), desc="Step 2: Fitting Lowess"):
        mask = df["ID"] == ID
        sc_df = df[mask]
        
        # Filter for valid data only
        valid_df = sc_df.dropna(subset=["Time (hours)", "Mtb Area Processed (\u00b5m)"])
        if len(valid_df) < 5: continue
        
        # Sort by time (required for Lowess)
        valid_df = valid_df.sort_values("Time (hours)")
        valid_idx = valid_df.index # Save indices to map back later!
        
        time = valid_df["Time (hours)"].values
        pop = valid_df["Mtb Area Processed (\u00b5m)"].values
        
        try:
            # Fit Lowess
            z = sm.nonparametric.lowess(endog=pop, exog=time, frac=frac)
            time_model = z[:, 0]
            area_model = z[:, 1]
            
            # Assign back using SPECIFIC INDICES
            df.loc[valid_idx, "Time Model (hours)"] = time_model
            df.loc[valid_idx, "Mtb Area Model (\u00b5m)"] = area_model
            
            # Calculate and assign R2
            score = r2_score(pop, area_model)
            df.loc[mask, "r2"] = score
            
        except Exception:
            continue
            
    return df


# --- 3. DOUBLING METRICS ---

def compute_doubling_metrics(df, min_area=1.92, r2_threshold=0.7, min_doubling_time=4.0):
    """
    Calculates doubling times but filters out biological impossibilities.
    
    New Parameters:
    - min_doubling_time: Any doubling time faster than this (e.g. 4 hours) is considered 
      noise/segmentation error and is excluded from the list.
    """
    df = df.copy()
    df["Doubling Amounts"] = None
    df["Doubling Times"] = None
    
    for ID in tqdm(df["ID"].unique(), desc="Doubling Metrics"):
        mask = df["ID"] == ID
        full_group_len = mask.sum()
        
        # Get valid model data
        sc_df = df[mask].dropna(subset=["Time Model (hours)", "Mtb Area Model (µm)"])
        if sc_df.empty: continue

        # Check R2
        r2 = sc_df["r2"].iloc[0]
        if r2 < r2_threshold: continue

        # Check Growth Range
        min_val = max(sc_df["Mtb Area Model (µm)"].min(), min_area)
        max_val = sc_df["Mtb Area Model (µm)"].max()
        if max_val <= min_val: continue

        # Generate Doubling Milestones
        N_series = []
        curr = min_val
        while curr <= max_val:
            N_series.append(curr)
            curr *= 2
        
        if len(N_series) < 2: continue

        # Calculate Times
        times = sc_df["Time Model (hours)"]
        vals = sc_df["Mtb Area Model (µm)"]
        
        doubling_idx = [np.abs(vals - target).idxmin() for target in N_series]
        
        # Calculate raw doubling intervals
        raw_times = times.loc[doubling_idx].values
        dt_intervals = np.diff(raw_times)
        
        # --- FILTERING STEP ---
        # We zip the intervals with the corresponding milestones to keep them aligned
        # If a doubling time is < 4 hours, we assume it's noise and drop it.
        
        valid_amounts = [N_series[0]] # Always keep start
        valid_intervals = []
        
        for i, dt in enumerate(dt_intervals):
            if dt >= min_doubling_time:
                valid_intervals.append(dt)
                valid_amounts.append(N_series[i+1])
            else:
                # If it doubled too fast (e.g. 0.5h), it's likely a segmentation jump.
                # We skip recording this specific interval.
                pass
        
        # If we filtered everything out, skip
        if not valid_intervals: continue

        # Safe Assignment
        df.loc[mask, "Doubling Amounts"] = pd.Series(
            [valid_amounts] * full_group_len, index=df[mask].index
        )
        df.loc[mask, "Doubling Times"] = pd.Series(
            [valid_intervals] * full_group_len, index=df[mask].index
        )

    return df
```

```python
import pandas as pd

def diagnose_dataframe_changes(df_old, df_new):
    print("--- Diagnostic Report ---")
    
    # 1. Check Dimensions
    print(f"Rows:    {len(df_old)} -> {len(df_new)} ({len(df_new) - len(df_old)})")
    print(f"Columns: {len(df_old.columns)} -> {len(df_new.columns)} ({len(df_new.columns) - len(df_old.columns)})")
    
    # 2. Identify Missing Columns
    old_cols = set(df_old.columns)
    new_cols = set(df_new.columns)
    missing_cols = old_cols - new_cols
    added_cols = new_cols - old_cols
    
    if missing_cols:
        print(f"\n[WARNING] {len(missing_cols)} Columns Lost:")
        print(list(missing_cols))
    else:
        print("\nNo columns were lost.")

    if added_cols:
        print(f"\n[INFO] {len(added_cols)} Columns Added:")
        print(list(added_cols))

    # 3. Check Memory Usage (Deep inspection)
    mem_old = df_old.memory_usage(deep=True).sum() / 1024**2
    mem_new = df_new.memory_usage(deep=True).sum() / 1024**2
    print(f"\nMemory Usage: {mem_old:.2f} MB -> {mem_new:.2f} MB")

    # 4. Check for Data Sparsity (NaNs)
    # If the new DF has significantly more NaNs, it may compress smaller on disk
    nan_old = df_old.isna().sum().sum()
    nan_new = df_new.isna().sum().sum()
    print(f"Total NaNs:   {nan_old} -> {nan_new} (Diff: {nan_new - nan_old:+})")

# Run the comparison
diagnose_dataframe_changes(df, df_merged)
```

```python
import napari
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import imageio.v3 as iio
import os
import io
import imageio  # This gives access to the v2 API
from skimage.transform import resize
import re  

# --- Configuration ---
VIDEO_DIR = '/mnt/OPERA3/Nathan/data/macrohet/macrohet_results/glimpse_store/glimpses/'
SAVE_PATH = '/mnt/OPERA3/Nathan/data/macrohet/macrohet_results/sc_df.ND0004_update.pkl'

# --- Ensure Style Defaults ---
sns.set_style("white")
plt.rcParams['font.family'] = 'Nimbus Sans'

class GlimpseAnnotator:
    def __init__(self, full_df, id_list):
        self.df = full_df         # Original DF with time-series lists
        self.id_list = id_list    # List of IDs to process
        self.current_idx = 0
        self.total = len(id_list)
        
        # Initialize Viewer
        self.viewer = napari.Viewer(title="Glimpse Annotator")
        
        # Setup bindings and load first sample
        self.setup_bindings()
        self.load_sample()

    def load_sample(self):
        """Loads video and plot, then resets time slider to 0."""
        if self.current_idx >= self.total:
            print("All samples processed!")
            return

        self.current_id = self.id_list[self.current_idx]
        
        # Clear existing layers
        self.viewer.layers.select_all()
        self.viewer.layers.remove_selected()
        
        print(f"--- Processing {self.current_idx + 1}/{self.total}: {self.current_id} ---")

        # 1. Attempt to Load Video
        video_path = os.path.join(VIDEO_DIR, f'{self.current_id}.mp4')
        video_loaded = False
        
        if os.path.exists(video_path):
            try:
                reader = imageio.get_reader(video_path, format='ffmpeg')
                video = np.stack([frame for frame in reader])
                reader.close()
                
                self.viewer.add_image(video, name=f'{self.current_id} Video')
                video_loaded = True
            except Exception as e:
                print(f"Warning: Could not read video for {self.current_id}. Error: {e}")
        else:
            print(f"Warning: Video file not found for {self.current_id}")

        # 2. Generate and Add Plot
        try:
            plot_img = self.create_plot_image(self.current_id)
            if plot_img is not None:
                self.viewer.add_image(plot_img, name='Growth Plot')
            else:
                print(f"No plot data generated for {self.current_id}")
        except Exception as e:
            print(f"Error generating plot for {self.current_id}: {e}")

        # 3. Layout and Time Reset
        self.viewer.grid.enabled = True
        self.viewer.grid.shape = (1, 2) if video_loaded else (1, 1)
        self.viewer.reset_view()
        
        # NEW: Reset Time Slider to 0 (t=0)
        if self.viewer.dims.ndim > 0:
            current_step = list(self.viewer.dims.current_step)
            current_step[0] = 0
            self.viewer.dims.current_step = tuple(current_step)

    def create_plot_image(self, ID):
        """Generates the Matplotlib graph with Doubling markers."""
        # 1. Filter and Sort
        sc_df = self.df[self.df['ID'] == ID].copy()
        if sc_df.empty: return None
        sc_df = sc_df.sort_values(by='Time Model (hours)')

        # Create Figure
        fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
        
        try:
            # 2. Extract Data & Metadata
            times = sc_df['Time Model (hours)']
            area_model = sc_df['Mtb Area Model (µm)']
            area_proc = sc_df['Mtb Area Processed (µm)']
            
            # Safe Metadata Extraction (Handle missing columns/NaNs)
            meta = sc_df.iloc[0]
            r2 = meta.get('r2', np.nan)
            strain = meta.get('Strain', 'Unknown')
            compound = meta.get('Compound', 'N/A')
            conc = meta.get('Concentration', 'N/A')
            
            # Check for Doubling Data (Lists)
            # These are stored as lists in the cell, e.g., [1.92, 3.84]
            d_amounts = meta.get('Doubling Amounts')
            d_times = meta.get('Doubling Times') # These are the intervals (e.g. 25h)
            
            # 3. Base Plot
            ax.plot(times, area_model, color='#d02c91', lw=2, label='Model')
            ax.scatter(times, area_proc, color='#1a9641', s=10, alpha=0.6, label='Data')
            
            # 4. Plot Doubling Lines (Crosshairs)
            if isinstance(d_amounts, list) and len(d_amounts) > 0:
                for amount in d_amounts:
                    # Find the exact time this amount was reached
                    # We interpolate to find Time (x) given Area (y)
                    # Note: x and y are swapped for np.interp
                    t_cross = np.interp(amount, area_model, times)
                    
                    # Draw Horizontal Line (Area)
                    ax.axhline(y=amount, color='gray', linestyle=':', alpha=0.5, lw=1)
                    # Draw Vertical Line (Time)
                    ax.axvline(x=t_cross, color='gray', linestyle=':', alpha=0.5, lw=1)
                    
                    # Optional: Add small text at the line
                    # ax.text(times.min(), amount, f"{amount:.1f}", fontsize=6, color='gray', va='bottom')

            # 5. Add Text Box for Doubling Intervals
            if isinstance(d_times, list) and len(d_times) > 0:
                # Format list: "20.5h, 18.2h"
                dt_str = ", ".join([f"{t:.1f}h" for t in d_times])
                info_text = f"Doubling Times:\n{dt_str}"
                
                # Place text box in bottom right
                props = dict(boxstyle='round', facecolor='white', alpha=0.8)
                ax.text(0.95, 0.05, info_text, transform=ax.transAxes, fontsize=9,
                        verticalalignment='bottom', horizontalalignment='right', bbox=props)

            # 6. Formatting
            # Update Title with Compound/Conc
            title_str = (f"ID: {ID}\n"
                         f"{strain} | {compound} {conc} | R2: {r2:.2f}")
            
            ax.set_title(title_str, fontsize=10)
            ax.legend(loc='upper left', fontsize=8)
            ax.set_xlabel("Time (hours)")
            ax.set_ylabel("Mtb Area (µm²)")
            sns.despine()
            
            # Convert to Image
            buf = io.BytesIO()
            fig.savefig(buf, format="png", bbox_inches='tight')
            plt.close(fig)
            buf.seek(0)
            img = imageio.v3.imread(buf, index=0)
            return img
            
        except Exception as e:
            print(f"Plotting error for {ID}: {e}")
            plt.close(fig)
            return None

    def next_sample(self, viewer):
        self.current_idx += 1
        self.load_sample()

    def prev_sample(self, viewer):
        if self.current_idx > 0:
            self.current_idx -= 1
            self.load_sample()

    # --- Classification Actions ---
    def update_origin(self, status):
        # Update the main DF
        self.df.loc[self.df['ID'] == self.current_id, 'mtb_origin'] = status
        print(f"Set {self.current_id} -> {status}")

    def mark_transfer(self, viewer): self.update_origin('Transfer')
    def mark_uptake(self, viewer): self.update_origin('Uptake')
    def mark_junk(self, viewer): self.update_origin('Junk')
    def mark_growth(self, viewer): self.update_origin('Growth')
    
    def mark_edge(self, viewer):
        self.df.loc[self.df['ID'] == self.current_id, 'Edge Status'] = True
        print(f"ID {self.current_id} marked as 'Edge Status'.")

    def save_df(self, viewer):
        print("Saving dataframe...")
        tmp_path = SAVE_PATH + '.tmp'
        self.df.to_pickle(tmp_path)
        os.replace(tmp_path, SAVE_PATH)
        print(f"Saved to {SAVE_PATH}")

    def split_track(self, viewer):
        """
        Splits the track with HEAVY debugging output to diagnose failures.
        """
        print("\n" + "="*40)
        print(f"DEBUG: Attempting Split on {self.current_id}")
        
        # 1. Get relative frame from slider
        try:
            current_frame = viewer.dims.current_step[0]
            print(f"DEBUG: Slider Frame: {current_frame}")
        except Exception as e:
            print(f"ERROR: Could not get frame from viewer: {e}")
            return
        
        # 2. Find Track Start
        track_df = self.df[self.df['ID'] == self.current_id]
        if track_df.empty:
            print(f"ERROR: ID {self.current_id} not found in DataFrame!")
            return
            
        start_time = track_df['Time (hours)'].min()
        end_time = track_df['Time (hours)'].max()
        split_time = start_time + (current_frame * 0.5)
        
        print(f"DEBUG: Track Start: {start_time:.2f}h | End: {end_time:.2f}h")
        print(f"DEBUG: Calculated Split Time: {split_time:.2f}h")
        
        if split_time > end_time:
            print("ERROR: Split time is AFTER the track ends. (Are you at the end of the slider?)")
            return
        if split_time < start_time:
            print("ERROR: Split time is BEFORE the track starts. (Calculation error?)")
            return

        # 3. Determine New ID Name
        try:
            root_id = re.sub(r'[a-z]+$', '', self.current_id)
            print(f"DEBUG: Root ID detected: '{root_id}'")
            
            # Find existing in the whole DF
            existing_ids = self.df[self.df['ID'].str.startswith(root_id)]['ID'].unique()
            print(f"DEBUG: Existing Family IDs: {existing_ids}")
            
            suffixes = [uid.replace(root_id, '') for uid in existing_ids]
            letters = [s for s in suffixes if s.isalpha()]
            
            if not letters:
                next_char = 'b'
            else:
                max_char = max(letters)
                next_char = chr(ord(max_char) + 1)
                
            new_id = root_id + next_char
            print(f"DEBUG: Assigned New ID: {new_id}")
            
        except Exception as e:
            print(f"ERROR: Failed during ID generation: {e}")
            return

        # 4. Apply Split
        mask = (self.df['ID'] == self.current_id) & (self.df['Time (hours)'] >= split_time)
        rows_affected = mask.sum()
        print(f"DEBUG: Rows matched for move: {rows_affected}")
        
        if rows_affected == 0:
            print("WARNING: No rows matched the split criteria. Aborting.")
            return

        # EXECUTE SPLIT
        self.df.loc[mask, 'ID'] = new_id
        print(f"SUCCESS: Moved {rows_affected} rows to {new_id}")
        
        # 5. Recalculate Metrics
        try:
            self.refresh_metrics_for_ids([self.current_id, new_id])
        except Exception as e:
            print(f"ERROR: Metric recalculation failed: {e}")
            # We continue anyway to at least show the split
        
        # 6. Refresh
        print("DEBUG: Reloading sample...")
        self.load_sample()
        print("="*40 + "\n")


    def refresh_metrics_for_ids(self, id_list):
        """
        Runs the pipeline on specific IDs.
        """
        print(f"   -> RECALC: Running pipeline for {id_list}")
        
        mask = self.df['ID'].isin(id_list)
        if mask.sum() == 0:
            print("   -> ERROR: No rows found for these IDs to recalculate.")
            return

        sub_df = self.df[mask].copy()
        
        # Pipeline
        try:
            # 1. Smooth
            sub_df = process_mtb_area_relaxed(sub_df, window=10, spike_threshold=10.0)
            
            # 2. Fit Lowess
            # Using robust version logic inline or calling the function if available
            sub_df = fit_lowess(sub_df, frac=0.25) 
            
            # 3. Metrics (with noise filter)
            sub_df = compute_doubling_metrics(sub_df, min_area=1.92, r2_threshold=0.7, min_doubling_time=4.0)
            
            # Update
            cols_to_update = [
                'Mtb Area Processed (µm)', 
                'Time Model (hours)', 
                'Mtb Area Model (µm)', 
                'r2', 
                'Doubling Amounts', 
                'Doubling Times'
            ]
            
            # Check if columns exist in sub_df (in case pipeline failed silently)
            available_cols = [c for c in cols_to_update if c in sub_df.columns]
            
            self.df.loc[mask, available_cols] = sub_df[available_cols]
            print("   -> RECALC: Complete.")
            
        except Exception as e:
            print(f"   -> RECALC ERROR: {e}")
            import traceback
            traceback.print_exc()

    # --- Update setup_bindings to include the new key ---
    def setup_bindings(self):
        self.viewer.bind_key('n', self.next_sample)
        self.viewer.bind_key('b', self.prev_sample)
        self.viewer.bind_key('s', self.save_df)
        self.viewer.bind_key('t', self.mark_transfer)
        self.viewer.bind_key('u', self.mark_uptake)
        self.viewer.bind_key('j', self.mark_junk)
        self.viewer.bind_key('g', self.mark_growth)
        self.viewer.bind_key('e', self.mark_edge)
        self.viewer.bind_key('k', self.split_track) # 'x' for cut/split


# --- Execution ---
if __name__ == "__main__":
    # 1. Get the list of IDs from your FILTERED subset
    target_ids = subset_df['ID'].unique()[654:]
    
    # 2. Pass the FULL df (for plotting) and the TARGET list (for iteration)
    app = GlimpseAnnotator(df, target_ids)
    
    print("Viewer launched.")
```

```python
import napari
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import imageio  # Using standard import to match your working script
import os
import io
import re
import statsmodels.api as sm
from sklearn.metrics import r2_score

# ==========================================
# PART 1: HELPER FUNCTIONS
# ==========================================

def smooth_and_fix_relaxed(area_series, window=10, spike_threshold=10.0):
    original_index = area_series.index
    area_series = area_series.reset_index(drop=True)
    rolling_mean = area_series.rolling(window=window, min_periods=1, center=True).mean()
    cleaned = area_series.copy()
    for i in range(1, len(cleaned) - 1):
        if cleaned.iloc[i] > spike_threshold * rolling_mean.iloc[i]:
            cleaned.iloc[i] = np.nan
        elif (cleaned.iloc[i] == 0 and cleaned.iloc[i - 1] > 0 and cleaned.iloc[i + 1] > 0):
            cleaned.iloc[i] = np.nan
    result = cleaned.interpolate(limit_direction='both')
    result.index = original_index
    return result

def process_mtb_area_relaxed(df, window=10, spike_threshold=10.0):
    df = df.copy()
    cleaned_series = df.groupby("ID")["Mtb Area (\u00b5m)"].transform(
        lambda x: smooth_and_fix_relaxed(x, window, spike_threshold)
    )
    df["Mtb Area Processed (\u00b5m)"] = cleaned_series
    return df

def fit_lowess(df, frac=0.25):
    df = df.copy()
    cols = ["Time Model (hours)", "Mtb Area Model (\u00b5m)", "r2"]
    for c in cols:
        if c not in df.columns: df[c] = np.nan
    for ID in df["ID"].unique():
        mask = df["ID"] == ID
        sc_df = df[mask]
        valid_df = sc_df.dropna(subset=["Time (hours)", "Mtb Area Processed (\u00b5m)"])
        if len(valid_df) < 5: continue
        valid_df = valid_df.sort_values("Time (hours)")
        valid_idx = valid_df.index 
        time = valid_df["Time (hours)"].values
        pop = valid_df["Mtb Area Processed (\u00b5m)"].values
        try:
            z = sm.nonparametric.lowess(endog=pop, exog=time, frac=frac)
            df.loc[valid_idx, "Time Model (hours)"] = z[:, 0]
            df.loc[valid_idx, "Mtb Area Model (\u00b5m)"] = z[:, 1]
            df.loc[mask, "r2"] = r2_score(pop, z[:, 1])
        except Exception: continue
    return df

def compute_doubling_metrics(df, min_area=1.92, r2_threshold=0.7, min_doubling_time=4.0):
    df = df.copy()
    df["Doubling Amounts"] = None
    df["Doubling Times"] = None
    for ID in df["ID"].unique():
        mask = df["ID"] == ID
        full_group_len = mask.sum()
        sc_df = df[mask].dropna(subset=["Time Model (hours)", "Mtb Area Model (\u00b5m)"])
        if sc_df.empty: continue
        if sc_df["r2"].iloc[0] < r2_threshold: continue
        min_val = max(sc_df["Mtb Area Model (\u00b5m)"].min(), min_area)
        max_val = sc_df["Mtb Area Model (\u00b5m)"].max()
        if max_val <= min_val: continue
        N_series = []
        curr = min_val
        while curr <= max_val:
            N_series.append(curr)
            curr *= 2
        if len(N_series) < 2: continue
        times = sc_df["Time Model (hours)"]
        vals = sc_df["Mtb Area Model (\u00b5m)"]
        doubling_idx = [np.abs(vals - target).idxmin() for target in N_series]
        dt_intervals = np.diff(times.loc[doubling_idx].values)
        valid_amounts = [N_series[0]]
        valid_intervals = []
        for i, dt in enumerate(dt_intervals):
            if dt >= min_doubling_time:
                valid_intervals.append(dt)
                valid_amounts.append(N_series[i+1])
        if not valid_intervals: continue
        df.loc[mask, "Doubling Amounts"] = pd.Series([valid_amounts] * full_group_len, index=df[mask].index)
        df.loc[mask, "Doubling Times"] = pd.Series([valid_intervals] * full_group_len, index=df[mask].index)
    return df

# ==========================================
# PART 2: THE ANNOTATOR CLASS
# ==========================================

class GlimpseAnnotator:
    def __init__(self, full_df, id_list, start_index=0, video_dir='./glimpses/', save_path='./verified_df.pkl'):
        self.df = full_df         
        self.id_list = id_list    
        self.video_dir = video_dir
        self.save_path = save_path
        self.current_idx = start_index
        self.total = len(id_list)
        self.viewer = napari.Viewer(title="Growth Profile Verifier")
        
        self.setup_bindings()
        
        print(f"🚀 Initializing at Index {self.current_idx}...")
        # Start directly
        self.load_sample()

    def setup_bindings(self):
        self.viewer.bind_key('n', self.next_sample)
        self.viewer.bind_key('b', self.prev_sample)
        self.viewer.bind_key('s', self.save_df)
        self.viewer.bind_key('k', self.split_track)
        self.viewer.bind_key('t', self.mark_transfer)
        self.viewer.bind_key('u', self.mark_uptake)
        self.viewer.bind_key('j', self.mark_junk)
        self.viewer.bind_key('g', self.mark_growth)
        self.viewer.bind_key('e', self.mark_edge)

    def load_sample(self):
        if self.current_idx >= self.total:
            print("All samples processed!")
            return

        self.current_id = self.id_list[self.current_idx]
        
        # Clear Layers
        if self.viewer.layers:
            self.viewer.layers.select_all()
            self.viewer.layers.remove_selected()
        
        print(f"--- Processing {self.current_idx}/{self.total}: {self.current_id} ---")

        video_path = os.path.join(self.video_dir, f'{self.current_id}.mp4')
        if not os.path.exists(video_path):
            print(f"⚠️ Missing Video. Skipping...")
            self.current_idx += 1
            if self.current_idx < self.total: self.load_sample()
            return

        layers_added = False

        # --- 1. LOAD VIDEO (Exact Logic from your Diagnostic) ---
        try:
            reader = imageio.get_reader(video_path, format='ffmpeg')
            # Using the exact list comp that worked for you
            video_data = np.stack([frame for frame in reader])
            reader.close()
            
            # Napari needs to know if it's RGB
            is_rgb = (video_data.ndim == 4 and video_data.shape[-1] in [3, 4])
            
            self.viewer.add_image(video_data, name=f'{self.current_id}', rgb=is_rgb)
            layers_added = True
            
        except Exception as e:
            print(f"❌ Error loading video data: {e}")

        # --- 2. LOAD PLOT ---
        try:
            plot_img = self.create_plot_image(self.current_id)
            if plot_img is not None:
                self.viewer.add_image(plot_img, name='Growth Profile', rgb=True)
                layers_added = True
        except Exception as e:
            print(f"❌ Error generating plot: {e}")

        # --- 3. GRID ---
        if layers_added:
            try:
                self.viewer.grid.enabled = True
                self.viewer.grid.shape = (1, 2)
                self.viewer.reset_view()
                
                # Reset Time Slider
                if self.viewer.dims.ndim > 0:
                    current_step = list(self.viewer.dims.current_step)
                    current_step[0] = 0
                    self.viewer.dims.current_step = tuple(current_step)
            except Exception as e:
                print(f"⚠️ Grid layout warning: {e}")
        else:
            print("⚠️ No layers loaded. Press 'n' to skip.")

    def create_plot_image(self, ID):
        sc_df = self.df[self.df['ID'] == ID].copy()
        sc_df = sc_df.dropna(subset=['Time Model (hours)', 'Mtb Area Model (µm)'])
        if sc_df.empty: return None
        sc_df = sc_df.sort_values(by='Time Model (hours)')

        meta = sc_df.iloc[0]
        strain = meta.get('Strain', '')
        compound = meta.get('Compound', '')
        conc = meta.get('Concentration', '')
        r2 = meta.get('r2', np.nan)
        d_amounts = meta.get('Doubling Amounts')
        if not isinstance(d_amounts, list): d_amounts = []
        
        fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
        colors = ['#1a9641', '#a6d96a', '#978897', '#d1d1ca', '#f1b6da', '#d02c91']
        
        ax.plot(sc_df['Time Model (hours)'], sc_df['Mtb Area Model (µm)'], label='Model', color=colors[-1], lw=2)
        ax.scatter(sc_df['Time Model (hours)'], sc_df['Mtb Area Processed (µm)'], label='Data', color=colors[0], s=10, alpha=0.6)

        if len(d_amounts) >= 2:
            doubling_indices = np.clip(np.searchsorted(sc_df['Mtb Area Model (µm)'], d_amounts), 0, len(sc_df) - 1)
            doubling_times = sc_df['Time Model (hours)'].iloc[doubling_indices].values
            
            for i, (t, amt) in enumerate(zip(doubling_times, d_amounts)):
                c = colors[i % len(colors)]
                ax.axvline(x=t, color=c, linestyle=':', alpha=0.6)
                ax.axhline(y=amt, color=c, linestyle=':', alpha=0.6)
                if i > 0:
                    prev_t = doubling_times[i-1]
                    ax.text(t, amt + (amt*0.02), f'{t - prev_t:.1f}h', ha='right', va='bottom', fontsize=7, color=c, weight='bold')
                    ax.text(t + 0.5, amt - (amt*0.1), f'Δ{amt - d_amounts[i-1]:.1f}µm²', fontsize=7, color=c, ha='left', va='top', rotation=270, alpha=0.7)

        title_str = f"ID: {ID}\n{strain} | {compound} {conc} | R2: {r2:.2f}"
        ax.set_title(title_str, fontsize=10)
        ax.legend(loc='upper left', fontsize=8)
        ax.set_xlabel("Time (hours)")
        ax.set_ylabel("Mtb Area (µm²)")
        sns.despine(offset=10)
        
        # Use simple imread on the buffer (Standard imageio)
        buf = io.BytesIO()
        fig.savefig(buf, format="png", bbox_inches='tight')
        plt.close(fig)
        buf.seek(0)
        return imageio.imread(buf)

    def split_track(self, viewer):
        try:
            curr_frame = viewer.dims.current_step[0]
            track_df = self.df[self.df['ID'] == self.current_id]
            if track_df.empty: return
            start_time = track_df['Time (hours)'].min()
            split_time = start_time + (curr_frame * 0.5)
            print(f"\n✂️ Splitting {self.current_id} at {split_time:.2f}h")
        except Exception: return

        root_id = re.sub(r'[a-z]+$', '', self.current_id)
        existing = self.df[self.df['ID'].str.startswith(root_id)]['ID'].unique()
        letters = [id.replace(root_id, '') for id in existing if id.replace(root_id, '').isalpha()]
        next_char = chr(ord(max(letters)) + 1) if letters else 'b'
        new_id = root_id + next_char

        mask = (self.df['ID'] == self.current_id) & (self.df['Time (hours)'] >= split_time)
        if mask.sum() == 0: return

        self.df.loc[mask, 'ID'] = new_id
        print(f"✅ Moved {mask.sum()} rows to new ID: {new_id}")
        self.refresh_metrics_for_ids([self.current_id, new_id])
        self.load_sample()

    def refresh_metrics_for_ids(self, id_list):
        print(f"🔄 Recalculating metrics for: {id_list}")
        mask = self.df['ID'].isin(id_list)
        if mask.sum() == 0: return
        sub_df = self.df[mask].copy()
        sub_df = process_mtb_area_relaxed(sub_df)
        sub_df = fit_lowess(sub_df)
        sub_df = compute_doubling_metrics(sub_df)
        cols = ['Mtb Area Processed (µm)', 'Time Model (hours)', 'Mtb Area Model (µm)', 'r2', 'Doubling Amounts', 'Doubling Times']
        valid_cols = [c for c in cols if c in sub_df.columns]
        self.df.loc[mask, valid_cols] = sub_df[valid_cols]
        print("✨ Metrics updated.")

    def next_sample(self, viewer):
        self.current_idx += 1
        self.load_sample()
    def prev_sample(self, viewer):
        if self.current_idx > 0: self.current_idx -= 1
        self.load_sample()
    def save_df(self, viewer):
        print(f"💾 Saving to {self.save_path}...")
        self.df.to_pickle(self.save_path)
        print("✅ Saved.")
    def update_status(self, col, val):
        self.df.loc[self.df['ID'] == self.current_id, col] = val
        print(f"🏷️ {self.current_id} -> {col}: {val}")
    def mark_transfer(self, v): self.update_status('mtb_origin', 'Transfer')
    def mark_uptake(self, v): self.update_status('mtb_origin', 'Uptake')
    def mark_junk(self, v): self.update_status('mtb_origin', 'Junk')
    def mark_growth(self, v): self.update_status('mtb_origin', 'Growth')
    def mark_edge(self, v): self.update_status('Edge Status', True)
```

```python
import napari
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import imageio.v2 as imageio
import os
import io
import re
import statsmodels.api as sm
from sklearn.metrics import r2_score

# ==========================================
# 1. DEFINE YOUR TARGET LIST
# ==========================================
final_target_ids = sorted(list({
    '1032.3.9.ND0004b', '1288.3.6.ND0004b', '1405.4.4.ND0004b', '1486.3.3.ND0004b',
    '157.3.5.ND0004b', '1791.3.10.ND0004b', '2180.3.10.ND0004b', '2488.3.3.ND0004b',
    '2686.3.10.ND0004b', '289.4.4.ND0004b', '291.4.4.ND0004b', '2933.5.5.ND0004b',
    '335.3.6.ND0004b', '3482.5.3.ND0004b', '3494.3.9.ND0004b', '3554.3.8.ND0004b',
    '3587.4.3.ND0004b', '3714.3.7.ND0004b', '391.3.3.ND0004b', '4019.3.5.ND0004b',
    '47.3.3.ND0004b', '52.3.7.ND0004b', '5331.3.7.ND0004b', '557.5.5.ND0004b',
    '597.3.8.ND0004b', '898.3.9.ND0004b'
}))

# ==========================================
# 2. THE SPLIT REVIEW CLASS
# ==========================================

class SplitReviewAnnotator:
    def __init__(self, full_df, id_list, video_dir='./glimpses/', save_path='./verified_split_df.pkl'):
        self.df = full_df         
        self.id_list = id_list    
        self.video_dir = video_dir
        self.save_path = save_path
        self.current_idx = 0
        self.total = len(id_list)
        
        # Initialize Viewer
        self.viewer = napari.Viewer(title="Split Track Reviewer")
        self.setup_bindings()
        
        print(f"🚀 Starting Review of {self.total} Split Tracks...")
        self.load_sample()

    def setup_bindings(self):
        self.viewer.bind_key('n', self.next_sample)
        self.viewer.bind_key('b', self.prev_sample)
        self.viewer.bind_key('s', self.save_df)
        
        # Simple classification keys
        self.viewer.bind_key('g', self.mark_growth)
        self.viewer.bind_key('j', self.mark_junk)

    def add_safe_rgb(self, data, name):
        """Safely adds RGB data, splitting channels if the GPU crashes."""
        if not isinstance(data, np.ndarray): data = np.array(data)
        is_rgb = (data.ndim >= 3 and data.shape[-1] in [3, 4])
        
        try:
            self.viewer.add_image(data, name=name, rgb=is_rgb)
            return True
        except ValueError as e:
            if "broadcast" in str(e) and is_rgb:
                print(f"   ⚠️ GPU Fix: Splitting channels for '{name}'...")
                self.viewer.add_image(data[..., 0], name=f"{name}_R", colormap='red', blending='additive')
                if data.shape[-1] > 1:
                    self.viewer.add_image(data[..., 1], name=f"{name}_G", colormap='green', blending='additive')
                if data.shape[-1] > 2:
                    self.viewer.add_image(data[..., 2], name=f"{name}_B", colormap='blue', blending='additive')
                return True
            else:
                print(f"   ❌ Add Image Failed: {e}")
                return False

    def load_sample(self):
        if self.current_idx >= self.total:
            print("All samples processed!")
            return

        self.current_id = self.id_list[self.current_idx]
        
        # Clear Layers
        if self.viewer.layers:
            self.viewer.layers.select_all()
            self.viewer.layers.remove_selected()
        
        # ---------------------------------------------------------
        # LOGIC: DETERMINE PARENT VIDEO ID
        # ---------------------------------------------------------
        # We strip any trailing letters (b, c, d...) to find the root
        root_id = re.sub(r'[a-z]+$', '', self.current_id)
        
        print(f"--- Reviewing {self.current_idx + 1}/{self.total} ---")
        print(f"    Target ID (Graph): {self.current_id}")
        print(f"    Parent ID (Video): {root_id}")

        video_path = os.path.join(self.video_dir, f'{root_id}.mp4')
        
        if not os.path.exists(video_path):
            print(f"⚠️ Parent Video Not Found: {video_path}")
            # Try exact match just in case
            video_path = os.path.join(self.video_dir, f'{self.current_id}.mp4')
            if not os.path.exists(video_path):
                print("   Skipping...")
                self.current_idx += 1
                if self.current_idx < self.total: self.load_sample()
                return

        layers_loaded = False

        # 1. LOAD PARENT VIDEO
        try:
            reader = imageio.get_reader(video_path, format='ffmpeg')
            video_data = np.stack([frame for frame in reader])
            reader.close()
            if self.add_safe_rgb(video_data, name=f'Video: {root_id}'):
                layers_loaded = True
        except Exception as e:
            print(f"❌ Error loading video: {e}")

        # 2. LOAD NEW PLOT (For the split 'b' ID)
        try:
            plot_img = self.create_plot_image(self.current_id)
            if plot_img is not None:
                if self.add_safe_rgb(plot_img, name=f'Graph: {self.current_id}'):
                    layers_loaded = True
            else:
                print(f"⚠️ No graph data found for {self.current_id}")
        except Exception as e:
            print(f"❌ Error loading plot: {e}")

        # 3. LAYOUT
        if layers_loaded:
            try:
                self.viewer.grid.enabled = True
                self.viewer.grid.shape = (1, 2)
                self.viewer.reset_view()
                if self.viewer.dims.ndim > 0:
                    current_step = list(self.viewer.dims.current_step)
                    current_step[0] = 0
                    self.viewer.dims.current_step = tuple(current_step)
            except Exception as e:
                print(f"⚠️ Layout warning: {e}")

    def create_plot_image(self, ID):
        """Generates plot for the SPECIFIC split ID."""
        sc_df = self.df[self.df['ID'] == ID].copy()
        sc_df = sc_df.dropna(subset=['Time Model (hours)', 'Mtb Area Model (µm)'])
        if sc_df.empty: return None
        sc_df = sc_df.sort_values(by='Time Model (hours)')

        meta = sc_df.iloc[0]
        # Get metrics
        strain = meta.get('Strain', '')
        compound = meta.get('Compound', '')
        r2 = meta.get('r2', np.nan)
        d_amounts = meta.get('Doubling Amounts')
        if not isinstance(d_amounts, list): d_amounts = []
        
        fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
        colors = ['#1a9641', '#a6d96a', '#978897', '#d1d1ca', '#f1b6da', '#d02c91']
        
        # Plot Curve
        ax.plot(sc_df['Time Model (hours)'], sc_df['Mtb Area Model (µm)'], label='Model', color=colors[-1], lw=2)
        ax.scatter(sc_df['Time Model (hours)'], sc_df['Mtb Area Processed (µm)'], label='Data', color=colors[0], s=10, alpha=0.6)

        # Plot Doublings
        if len(d_amounts) >= 2:
            doubling_indices = np.clip(np.searchsorted(sc_df['Mtb Area Model (µm)'], d_amounts), 0, len(sc_df) - 1)
            doubling_times = sc_df['Time Model (hours)'].iloc[doubling_indices].values
            
            for i, (t, amt) in enumerate(zip(doubling_times, d_amounts)):
                c = colors[i % len(colors)]
                ax.axvline(x=t, color=c, linestyle=':', alpha=0.6)
                ax.axhline(y=amt, color=c, linestyle=':', alpha=0.6)
                if i > 0:
                    prev_t = doubling_times[i-1]
                    ax.text(t, amt + (amt*0.02), f'{t - prev_t:.1f}h', ha='right', va='bottom', fontsize=7, color=c, weight='bold')

        title_str = f"SPLIT ID: {ID}\n{strain} | {compound} | R2: {r2:.2f}"
        ax.set_title(title_str, fontsize=10, color='red', weight='bold') # Red title for split awareness
        ax.legend(loc='upper left', fontsize=8)
        ax.set_xlabel("Time (hours)")
        ax.set_ylabel("Mtb Area (µm²)")
        sns.despine(offset=10)
        
        buf = io.BytesIO()
        fig.savefig(buf, format="png", bbox_inches='tight')
        plt.close(fig)
        buf.seek(0)
        return imageio.imread(buf)

    def next_sample(self, viewer):
        self.current_idx += 1
        self.load_sample()

    def prev_sample(self, viewer):
        if self.current_idx > 0: self.current_idx -= 1
        self.load_sample()
        
    def save_df(self, viewer):
        print(f"💾 Saving to {self.save_path}...")
        self.df.to_pickle(self.save_path)
        print("✅ Saved.")

    def mark_growth(self, v): 
        self.df.loc[self.df['ID'] == self.current_id, 'mtb_origin'] = 'Growth'
        print(f"🏷️ {self.current_id} -> Growth")
        
    def mark_junk(self, v):
        self.df.loc[self.df['ID'] == self.current_id, 'mtb_origin'] = 'Junk'
        print(f"🏷️ {self.current_id} -> Junk")

# ==========================================
# 3. LAUNCH APP
# ==========================================
# Pass your main DataFrame (df) which contains the new split rows
app = SplitReviewAnnotator(
    df, 
    final_target_ids, 
    video_dir='/mnt/OPERA3/Nathan/data/macrohet/macrohet_results/glimpse_store/glimpses/'
)
```

```python
import napari
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import imageio.v2 as imageio
import os
import io
import re
import statsmodels.api as sm
from sklearn.metrics import r2_score

# ==========================================
# PART 1: HELPER FUNCTIONS
# ==========================================

def smooth_and_fix_relaxed(area_series, window=10, spike_threshold=10.0):
    original_index = area_series.index
    area_series = area_series.reset_index(drop=True)
    rolling_mean = area_series.rolling(window=window, min_periods=1, center=True).mean()
    cleaned = area_series.copy()
    for i in range(1, len(cleaned) - 1):
        if cleaned.iloc[i] > spike_threshold * rolling_mean.iloc[i]:
            cleaned.iloc[i] = np.nan
        elif (cleaned.iloc[i] == 0 and cleaned.iloc[i - 1] > 0 and cleaned.iloc[i + 1] > 0):
            cleaned.iloc[i] = np.nan
    result = cleaned.interpolate(limit_direction='both')
    result.index = original_index
    return result

def process_mtb_area_relaxed(df, window=10, spike_threshold=10.0):
    df = df.copy()
    cleaned_series = df.groupby("ID")["Mtb Area (\u00b5m)"].transform(
        lambda x: smooth_and_fix_relaxed(x, window, spike_threshold)
    )
    df["Mtb Area Processed (\u00b5m)"] = cleaned_series
    return df

def fit_lowess(df, frac=0.25):
    df = df.copy()
    cols = ["Time Model (hours)", "Mtb Area Model (\u00b5m)", "r2"]
    for c in cols:
        if c not in df.columns: df[c] = np.nan
    for ID in df["ID"].unique():
        mask = df["ID"] == ID
        sc_df = df[mask]
        valid_df = sc_df.dropna(subset=["Time (hours)", "Mtb Area Processed (\u00b5m)"])
        if len(valid_df) < 5: continue
        valid_df = valid_df.sort_values("Time (hours)")
        valid_idx = valid_df.index 
        time = valid_df["Time (hours)"].values
        pop = valid_df["Mtb Area Processed (\u00b5m)"].values
        try:
            z = sm.nonparametric.lowess(endog=pop, exog=time, frac=frac)
            df.loc[valid_idx, "Time Model (hours)"] = z[:, 0]
            df.loc[valid_idx, "Mtb Area Model (\u00b5m)"] = z[:, 1]
            df.loc[mask, "r2"] = r2_score(pop, z[:, 1])
        except Exception: continue
    return df

def compute_doubling_metrics(df, min_area=1.92, r2_threshold=0.7, min_doubling_time=4.0):
    df = df.copy()
    df["Doubling Amounts"] = None
    df["Doubling Times"] = None
    for ID in df["ID"].unique():
        mask = df["ID"] == ID
        full_group_len = mask.sum()
        sc_df = df[mask].dropna(subset=["Time Model (hours)", "Mtb Area Model (\u00b5m)"])
        if sc_df.empty: continue
        if sc_df["r2"].iloc[0] < r2_threshold: continue
        min_val = max(sc_df["Mtb Area Model (\u00b5m)"].min(), min_area)
        max_val = sc_df["Mtb Area Model (\u00b5m)"].max()
        if max_val <= min_val: continue
        N_series = []
        curr = min_val
        while curr <= max_val:
            N_series.append(curr)
            curr *= 2
        if len(N_series) < 2: continue
        times = sc_df["Time Model (hours)"]
        vals = sc_df["Mtb Area Model (\u00b5m)"]
        doubling_idx = [np.abs(vals - target).idxmin() for target in N_series]
        dt_intervals = np.diff(times.loc[doubling_idx].values)
        valid_amounts = [N_series[0]]
        valid_intervals = []
        for i, dt in enumerate(dt_intervals):
            if dt >= min_doubling_time:
                valid_intervals.append(dt)
                valid_amounts.append(N_series[i+1])
        if not valid_intervals: continue
        df.loc[mask, "Doubling Amounts"] = pd.Series([valid_amounts] * full_group_len, index=df[mask].index)
        df.loc[mask, "Doubling Times"] = pd.Series([valid_intervals] * full_group_len, index=df[mask].index)
    return df

# ==========================================
# PART 2: THE ROBUST CLASS (Split-Channel Fix)
# ==========================================

class GlimpseAnnotator:
    def __init__(self, full_df, id_list, start_index=0, video_dir='./glimpses/', save_path='./verified_df.pkl'):
        self.df = full_df         
        self.id_list = id_list    
        self.video_dir = video_dir
        self.save_path = save_path
        self.current_idx = start_index
        self.total = len(id_list)
        
        # Initialize Viewer
        self.viewer = napari.Viewer(title="Growth Profile Verifier")
        self.setup_bindings()
        
        print(f"🚀 Initializing at Index {self.current_idx}...")
        self.load_sample()

    def setup_bindings(self):
        self.viewer.bind_key('n', self.next_sample)
        self.viewer.bind_key('b', self.prev_sample)
        self.viewer.bind_key('s', self.save_df)
        self.viewer.bind_key('k', self.split_track)
        self.viewer.bind_key('t', self.mark_transfer)
        self.viewer.bind_key('u', self.mark_uptake)
        self.viewer.bind_key('j', self.mark_junk)
        self.viewer.bind_key('g', self.mark_growth)
        self.viewer.bind_key('e', self.mark_edge)

    def add_safe_rgb(self, data, name):
        """
        Attempts to load RGB data normally. 
        If it crashes due to the (3,) (2,) bug, it splits channels.
        """
        # Ensure data is numpy array
        if not isinstance(data, np.ndarray):
            data = np.array(data)

        # Detect if actually RGB
        is_rgb = (data.ndim >= 3 and data.shape[-1] in [3, 4])
        
        try:
            # 1. Try Normal Load
            self.viewer.add_image(data, name=name, rgb=is_rgb)
            return True
            
        except ValueError as e:
            if "broadcast" in str(e) and is_rgb:
                print(f"   ⚠️ Bypass: Splitting channels for '{name}' to avoid crash...")
                
                # 2. Fallback: Split Channels (Additive Blending)
                # Channel 0: Red
                self.viewer.add_image(
                    data[..., 0], name=f"{name}_R", colormap='red', blending='additive'
                )
                # Channel 1: Green
                if data.shape[-1] > 1:
                    self.viewer.add_image(
                        data[..., 1], name=f"{name}_G", colormap='green', blending='additive'
                    )
                # Channel 2: Blue
                if data.shape[-1] > 2:
                    self.viewer.add_image(
                        data[..., 2], name=f"{name}_B", colormap='blue', blending='additive'
                    )
                return True
            else:
                # Actual other error
                print(f"   ❌ Add Image Failed: {e}")
                return False

    def load_sample(self):
        if self.current_idx >= self.total:
            print("All samples processed!")
            return

        self.current_id = self.id_list[self.current_idx]
        
        # Clear Layers
        if self.viewer.layers:
            self.viewer.layers.select_all()
            self.viewer.layers.remove_selected()
        
        print(f"--- Processing {self.current_idx}/{self.total}: {self.current_id} ---")

        video_path = os.path.join(self.video_dir, f'{self.current_id}.mp4')
        if not os.path.exists(video_path):
            print(f"⚠️ Missing Video. Skipping...")
            self.current_idx += 1
            if self.current_idx < self.total: self.load_sample()
            return

        layers_loaded = False

        # --- 1. LOAD VIDEO ---
        try:
            reader = imageio.get_reader(video_path, format='ffmpeg')
            video_data = np.stack([frame for frame in reader])
            reader.close()
            
            # Use the SAFE loader
            if self.add_safe_rgb(video_data, name=f'{self.current_id}'):
                layers_loaded = True
            
        except Exception as e:
            print(f"❌ Error loading video: {e}")

        # --- 2. LOAD PLOT ---
        try:
            plot_img = self.create_plot_image(self.current_id)
            if plot_img is not None:
                # Use the SAFE loader (plots are also RGB)
                if self.add_safe_rgb(plot_img, name='Growth Profile'):
                    layers_loaded = True
        except Exception as e:
            print(f"❌ Error loading plot: {e}")

        # --- 3. LAYOUT ---
        if layers_loaded:
            try:
                self.viewer.grid.enabled = True
                self.viewer.grid.shape = (1, 2)
                self.viewer.reset_view()
                
                if self.viewer.dims.ndim > 0:
                    current_step = list(self.viewer.dims.current_step)
                    current_step[0] = 0
                    self.viewer.dims.current_step = tuple(current_step)
            except Exception as e:
                print(f"⚠️ Grid layout warning: {e}")
        else:
            print("⚠️ No layers loaded. Press 'n' to skip.")

    def create_plot_image(self, ID):
        sc_df = self.df[self.df['ID'] == ID].copy()
        sc_df = sc_df.dropna(subset=['Time Model (hours)', 'Mtb Area Model (µm)'])
        if sc_df.empty: return None
        sc_df = sc_df.sort_values(by='Time Model (hours)')

        meta = sc_df.iloc[0]
        strain = meta.get('Strain', '')
        compound = meta.get('Compound', '')
        conc = meta.get('Concentration', '')
        r2 = meta.get('r2', np.nan)
        d_amounts = meta.get('Doubling Amounts')
        if not isinstance(d_amounts, list): d_amounts = []
        
        fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
        colors = ['#1a9641', '#a6d96a', '#978897', '#d1d1ca', '#f1b6da', '#d02c91']
        
        ax.plot(sc_df['Time Model (hours)'], sc_df['Mtb Area Model (µm)'], label='Model', color=colors[-1], lw=2)
        ax.scatter(sc_df['Time Model (hours)'], sc_df['Mtb Area Processed (µm)'], label='Data', color=colors[0], s=10, alpha=0.6)

        if len(d_amounts) >= 2:
            doubling_indices = np.clip(np.searchsorted(sc_df['Mtb Area Model (µm)'], d_amounts), 0, len(sc_df) - 1)
            doubling_times = sc_df['Time Model (hours)'].iloc[doubling_indices].values
            
            for i, (t, amt) in enumerate(zip(doubling_times, d_amounts)):
                c = colors[i % len(colors)]
                ax.axvline(x=t, color=c, linestyle=':', alpha=0.6)
                ax.axhline(y=amt, color=c, linestyle=':', alpha=0.6)
                if i > 0:
                    prev_t = doubling_times[i-1]
                    ax.text(t, amt + (amt*0.02), f'{t - prev_t:.1f}h', ha='right', va='bottom', fontsize=7, color=c, weight='bold')
                    ax.text(t + 0.5, amt - (amt*0.1), f'Δ{amt - d_amounts[i-1]:.1f}µm²', fontsize=7, color=c, ha='left', va='top', rotation=270, alpha=0.7)

        title_str = f"ID: {ID}\n{strain} | {compound} {conc} | R2: {r2:.2f}"
        ax.set_title(title_str, fontsize=10)
        ax.legend(loc='upper left', fontsize=8)
        ax.set_xlabel("Time (hours)")
        ax.set_ylabel("Mtb Area (µm²)")
        sns.despine(offset=10)
        
        buf = io.BytesIO()
        fig.savefig(buf, format="png", bbox_inches='tight')
        plt.close(fig)
        buf.seek(0)
        return imageio.imread(buf)

    def split_track(self, viewer):
        try:
            curr_frame = viewer.dims.current_step[0]
            track_df = self.df[self.df['ID'] == self.current_id]
            if track_df.empty: return
            start_time = track_df['Time (hours)'].min()
            split_time = start_time + (curr_frame * 0.5)
            print(f"\n✂️ Splitting {self.current_id} at {split_time:.2f}h")
        except Exception: return

        root_id = re.sub(r'[a-z]+$', '', self.current_id)
        existing = self.df[self.df['ID'].str.startswith(root_id)]['ID'].unique()
        letters = [id.replace(root_id, '') for id in existing if id.replace(root_id, '').isalpha()]
        next_char = chr(ord(max(letters)) + 1) if letters else 'b'
        new_id = root_id + next_char

        mask = (self.df['ID'] == self.current_id) & (self.df['Time (hours)'] >= split_time)
        if mask.sum() == 0: return

        self.df.loc[mask, 'ID'] = new_id
        print(f"✅ Moved {mask.sum()} rows to new ID: {new_id}")
        self.refresh_metrics_for_ids([self.current_id, new_id])
        self.load_sample()

    def refresh_metrics_for_ids(self, id_list):
        print(f"🔄 Recalculating metrics for: {id_list}")
        mask = self.df['ID'].isin(id_list)
        if mask.sum() == 0: return
        sub_df = self.df[mask].copy()
        sub_df = process_mtb_area_relaxed(sub_df)
        sub_df = fit_lowess(sub_df)
        sub_df = compute_doubling_metrics(sub_df)
        cols = ['Mtb Area Processed (µm)', 'Time Model (hours)', 'Mtb Area Model (µm)', 'r2', 'Doubling Amounts', 'Doubling Times']
        valid_cols = [c for c in cols if c in sub_df.columns]
        self.df.loc[mask, valid_cols] = sub_df[valid_cols]
        print("✨ Metrics updated.")

    def next_sample(self, viewer):
        self.current_idx += 1
        self.load_sample()
    def prev_sample(self, viewer):
        if self.current_idx > 0: self.current_idx -= 1
        self.load_sample()
    def save_df(self, viewer):
        print(f"💾 Saving to {self.save_path}...")
        self.df.to_pickle(self.save_path)
        print("✅ Saved.")
    def update_status(self, col, val):
        self.df.loc[self.df['ID'] == self.current_id, col] = val
        print(f"🏷️ {self.current_id} -> {col}: {val}")
    def mark_transfer(self, v): self.update_status('mtb_origin', 'Transfer')
    def mark_uptake(self, v): self.update_status('mtb_origin', 'Uptake')
    def mark_junk(self, v): self.update_status('mtb_origin', 'Junk')
    def mark_growth(self, v): self.update_status('mtb_origin', 'Growth')
    def mark_edge(self, v): self.update_status('Edge Status', True)

```

```python
df = pd.read_pickle('/mnt/OPERA3/Nathan/data/macrohet/macrohet_results/dfs/sc_df.ND0004_update.pkl')
# 1. Identify valid IDs (Duration > 35 hours)
#    We group by ID and check the maximum time recorded for each.
max_times = df.groupby('ID')['Time (hours)'].max()
valid_ids = max_times[max_times > 35].index

# 2. Apply filters (Metadata + Duration)
subset_df = df[
    (df['Experiment ID'] == 'ND0004') 
    & (df['Concentration'] != 'EC99') 
    & (df['Strain'] != 'UNI')
    & (df['r2'] >= 0.7)
    & (df['ID'].isin(valid_ids)) 
]

# 3. Drop duplicates to get one row per ID
#    (Assuming 'Doubling Times' is repeated in every row or you only need metadata now)
subset_df = subset_df.drop_duplicates(subset=['ID'])

# 2. If 'Doubling Times' is a list (e.g., [20, 45]), this expands it into 
#    multiple rows, one for each value, repeating the other metadata (ID, Strain, etc.).
subset_df = subset_df.explode('Doubling Times')

# 3. Converts 'Doubling Times' column to floats. 'coerce' turns non-numeric
#    strings/errors into NaN (Not a Number) to prevent the code from crashing.
subset_df['Doubling Times'] = pd.to_numeric(subset_df['Doubling Times'], errors='coerce')

# 4. Removes rows where 'Doubling Times' is NaN (resulting from the step above).
subset_df = subset_df[subset_df['Doubling Times'].notna()]

# 5. Filters out any zero or negative doubling times (we're only interested in growing populations).
subset_df = subset_df[subset_df['Doubling Times'] > 0]

# 6. Reorders the entire DataFrame so the fastest-growing cells (smallest doubling times) 
#    are at the top.
subset_df = subset_df.sort_values('Doubling Times', ascending=True)

# 7. Final deduplication: Since the data is sorted by shortest time, keeping 'first' 
#    ensures you retain the single fastest growth record for every unique 'ID'.
subset_df = subset_df.drop_duplicates(subset=['ID'], keep='first')
```
