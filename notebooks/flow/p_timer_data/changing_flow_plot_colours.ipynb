```python
import numpy as np
from PIL import Image
import os

# --- Configuration ---
input_path = os.path.expanduser('~/Downloads/pza.png')
# Changed output name slightly to indicate transparency
output_path = os.path.expanduser('~/Downloads/image_recoloured_pza.png')
target_hex = '3e0c52'#4a4c7e'
# Threshold for what counts as "background white" (higher = stricter)
white_threshold = 250


# --- Process ---

# 1. Convert hex to RGBA integer tuple (Red, Green, Blue, Alpha)
h = target_hex.lstrip('#')
rgb = tuple(int(h[i:i+2], 16) for i in (0, 2, 4))
# Append 255 to make the target colour fully opaque
target_rgba = rgb + (255,)

# 2. Load image and force convert to RGBA (4 channels)
img = Image.open(input_path).convert('RGBA')
data = np.array(img)

# 3. Create masks based on color, ignoring the existing alpha channel for now
# Isolate just the RGB channels to check for whiteness
rgb_data = data[:, :, :3]

# Find pixels where ALL (np.all) RGB values are above the threshold (i.e., white background)
is_background_mask = np.all(rgb_data >= white_threshold, axis=2)

# Find the inverse: pixels that are NOT background (i.e., your data)
is_foreground_mask = ~is_background_mask

# 4. Create a brand new empty image array filled with zeros
# Zeros in RGBA means completely transparent black (0,0,0,0)
new_data = np.zeros_like(data)

# 5. Fill only the foreground areas in the new image with your target colour
new_data[is_foreground_mask] = target_rgba

# 6. Save result explicitly as PNG (needed for transparency support)
result = Image.fromarray(new_data, 'RGBA')
result.save(output_path, format='PNG')
print(f"Processed with transparency: {output_path}")
```
